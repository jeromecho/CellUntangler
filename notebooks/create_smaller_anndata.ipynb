```python
import scanpy as sc

adata = sc.read_h5ad("../../../data/HGSOC/ALL_CELLS/all_cells_1p.h5ad")
```

```python
adata
```

```python
import numpy as np

n_total = adata.n_obs  # total number of cells
fraction = 0.001      # 0.05% expressed as a fraction
n_sample = max(1, int(n_total * fraction))  # ensure at least 1 cell

print(f"Sampling {n_sample} cells out of {n_total}")
```

```python
# Random cell indices
np.random.seed(42)  # for reproducibility
sample_indices = np.random.choice(n_total, size=n_sample, replace=False)
```

```python
adata_sub = adata[sample_indices, :].copy()
```

```python
adata_sub.write("../../../data/HGSOC/ALL_CELLS/all_cells_0001p.h5ad")
```
